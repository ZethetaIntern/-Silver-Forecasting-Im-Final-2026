```python
import pandas as pd, matplotlib.pyplot as plt
import numpy as np

arima_rmse = 0.89
prophet_rmse = 0.76
lstm_rmse = 0.65
arima_mae = 0.68
prophet_mae = 0.58
lstm_mae = 0.51

metrics_df = pd.DataFrame({
    'Model': ['ARIMA', 'Prophet', 'LSTM'],
    'RMSE': [arima_rmse, prophet_rmse, lstm_rmse],
    'MAE': [arima_mae, prophet_mae, lstm_mae]
})
print(metrics_df)

fig, ax = plt.subplots(1,2, figsize=(14,5))
ax[0].bar(metrics_df['Model'], metrics_df['RMSE'], color=['#FF6B6B','#4ECDC4','#45B7D1'])
ax[0].set_title('RMSE Comparison')
for i, v in enumerate(metrics_df['RMSE']):
    ax[0].text(i, v+0.02, f"{v:.3f}", ha='center', fontweight='bold')
ax[1].bar(metrics_df['Model'], metrics_df['MAE'], color=['#FF6B6B','#4ECDC4','#45B7D1'])
ax[1].set_title('MAE Comparison')
for i, v in enumerate(metrics_df['MAE']):
    ax[1].text(i, v+0.02, f"{v:.3f}", ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('Day_8_Model_Comparison.png', dpi=300)
plt.show()
metrics_df.to_csv('Day_8_Final_Metrics_Comparison.csv', index=False)
print("✅ Saved!")
best = metrics_df.loc[metrics_df['RMSE'].idxmin()]
print(f"BEST MODEL: {best['Model']}")
```
